## Problem 1 – Dataset Inspection

```python
# 1. Display the first 10 rows
print(df.head(10))

# 2. Find the number of rows and columns
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# 3. Identify all column names
print(df.columns)

# 4. Determine the data type of each column
print(df.dtypes)

# 5. Display descriptive statistics
print(df.describe())
```

## Problem 2 – Data Filtering

```python
# 1. Display first class passengers
print(df[df["Pclass"] == 1])

# 2. Find passengers younger than 18
print(df[df["Age"] < 18])

# 3. Find male passengers who survived
print(df[(df["Sex"] == "male") & (df["Survived"] == 1)])

# 4. Find passengers who paid a fare greater than 100
print(df[df["Fare"] > 100])
```

## Problem 3 – Sorting and Aggregation

```python
# 1. Display the 10 passengers who paid the highest fares
print(df.sort_values(
    by="Fare",
    ascending=False
    ).head(10))

# 2. Calculate the average passenger age
print("Average passenger age:", df["Age"].mean())

# 3. Count male and female passengers
print(df["Sex"].value_counts())

# 4. Calculate average fare for each passenger class
print(df.groupby("Pclass")["Fare"].mean())

# 5. Calculate survival rate for each passenger class
print(df.groupby("Pclass")["Survived"].mean())
```

# Final Laboratory Challenge

```python
# 1. Total number of passengers
print("Total number of passengers:", df.shape[0])

# 2. Average passenger age
print("Average passenger age:", df["Age"].mean())

# 3. Passenger class with the highest number of passengers
print("Number of passenger in each class:")
print(df["Pclass"].value_counts())

# 4. Average fare paid by male and female passengers
print("Average fare by gender:")
print(df.groupby("Sex")["Fare"].mean())

# 5. Survival rate by gender
print("Survival rate by gender:")
print(df.groupby("Sex")["Survived"].mean())

# 6. Survival rate across passenger classes
print("Survival rate by passenger class:")
print(df.groupby("Pclass")["Survived"].mean())
```
