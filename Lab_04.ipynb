```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("train.csv")
```

## Problem 1

```python
print(df.head())

print(df.info())

print(df.dtypes)

print(df.select_dtypes(include=['int64','float']).columns)
```

## Problem 2

```python
avrg_fare = df.groupby('Pclass')['Fare'].mean()

plt.plot(avrg_fare.index, avrg_fare.values)

plt.title("Average Fare Across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Average Fare")

plt.show()
```

## Problem 3


```python
plt.scatter(df['Age'], df['Fare'])

plt.title("Relationship Between Passenger Age and Fare")
plt.xlabel("Passenger Age")
plt.ylabel("Fare")
plt.show()
```

## problem 4


```python
gender_count = df['Sex'].value_counts()

plt.bar(gender_count.index, gender_count.values)

plt.title("Number of Male and Female Passenger")
plt.xlabel("Gender")
plt.ylabel("No of Passengers")
plt.show()
```

## problem 5


```python
plt.hist(df['Age'].dropna(), bins=10)

plt.title("Distribution of passenger Age")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

## problem 6

```python
survival_count = df['Survived'].value_counts()

plt.pie(
    survival_count,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)
plt.title("Passengers Who Survived")
plt.show()
```

## problem 7

```python
fig, ax = plt.subplots(1, 2,figsize=(15, 5))

ax[0].hist(df['Age'].dropna(), bins=10)
ax[0].set_title("Distribution of Passenger Ages")
ax[0].set_xlabel("Age")
ax[0].set_ylabel("Frequency")

class_count = df['Pclass'].value_counts().sort_index()

ax[1].bar(class_count.index, class_count.values)
ax[1].set_title("No of Passengers in Each Class")
ax[1].set_xlabel("Passenger Class")
ax[1].set_ylabel("Number of Passenger")

plt.tight_layout()
plt.show()
```

## problem 8


```python
fig, ax = plt.subplots(1, 3, figsize=(15, 5))

gender_count = df['Sex'].value_counts()

ax[0].bar(gender_count.index, gender_count.values)
ax[0].set_title("Passengers by Gender")
ax[0].set_xlabel("Gender")
ax[0].set_ylabel("Number of Passengers")

ax[1].hist(df['Fare'], bins=10)
ax[1].set_title("Distribution of Passenger Fare")
ax[1].set_xlabel("Fare")
ax[1].set_ylabel("Frequency")

survival_count = df['Survived'].value_counts()

ax[2].pie(
    survival_count,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)
ax[2].set_title("Passenger Survival Proportion")

plt.tight_layout()
plt.show()
```
